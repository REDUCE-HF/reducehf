```python
df = pd.read_csv("../../output/dataset_wp4.csv.gz", compression='gzip')
```

```python
date_columns = [col for col in df.columns if "_date" in col]
df[date_columns] = df[date_columns].apply(pd.to_datetime)
```

## Reduce data

```python
main_cols = [
    "patient_id",
    "sex",
    "ethnicity_cat",
    "index_date",
    "birth_date",
    "sysbp_value",
    "bmi_value",
    "last_cholesterol_value",
    "last_hdl_cholesterol_value",
    "last_hba1c_value",
    "smoking",
    "last_hypertension_date_med",
    "last_diabetes_medication_date",
    "hf_diagnosis_date"
]
df_main = df[main_cols]
```

```python
print (df_main.count())
```

## Derive variables

```python
#Age
df_main["age"] = (df_main["index_date"] - df_main["birth_date"]).dt.days / 365.25
df_main["bp_treatment"] = df_main["last_hypertension_date_med"].notna().astype(int)
df_main["dm_treatment"] = df_main["last_diabetes_medication_date"].notna().astype(int)
print(df_main[["age", "bp_treatment", "dm_treatment"]].describe())
```
