```python
%load_ext autoreload
%autoreload 2

import os
import pandas as pd
import numpy as np
import warnings

warnings.filterwarnings("ignore")

from config import OUTPUT_DIR

from pcp_parameters import (
    PCP_MODELS,
    calculate_pcp_hf_risk,
)

from validation_helpers import (
    evaluate_prediction_metrics,
)
```

## 1. Load Data

```python

# Load prepared validation dataset
DATA_VALIDATION_PATH = os.path.join(
    OUTPUT_DIR,
    "DATA_VALIDATION_FULL.csv.gz"
)

df_validation = pd.read_csv(
    DATA_VALIDATION_PATH,
    compression="gzip"
)
```

## 2. Run Models

```python
# White male
df_white_male = df_validation[
    (df_validation["sex"] == 1)
    & (df_validation["ethnicity_white"] == 1)
].copy()

df_white_male = calculate_pcp_hf_risk(
    df_white_male,
    PCP_MODELS[("white", "male")]
)
```

## 3. Performance Assessment

```python
results_white_male = evaluate_prediction_metrics(
    y_true=df_white_male["hf_diagnosis"],
    y_prob=df_white_male["pcp_hf_prediction"],
    threshold=0.10
)

results_white_male_df = pd.DataFrame.from_dict(
    results_white_male,
    orient="index",
    columns=["Value"]
)

print(results_white_male_df)
```
